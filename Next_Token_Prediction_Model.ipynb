```python

# Development of a Transformer-Based Next-Token Prediction Model
# Using Basic Self-Attention

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Embedding, MultiHeadAttention,
    LayerNormalization, Dense, Dropout, GlobalAveragePooling1D
)

# 1. Set random seeds
np.random.seed(42)
tf.random.set_seed(42)

# 2. Create a sample text dataset
sentences = [
    "the cat sits on the mat",
    "the dog sits on the floor",
    "the cat eats the food",
    "the dog eats the food",
    "the cat runs in the garden",
    "the dog runs in the park",
    "a cat is a small animal",
    "a dog is a friendly animal",
    "the sun shines in the sky",
    "the moon shines at night",
    "machine learning is a part of artificial intelligence",
    "deep learning uses neural networks",
    "a transformer model uses self attention",
    "self attention learns relationships between tokens",
    "a language model predicts the next token",
    "the model learns patterns from text",
    "artificial intelligence helps solve problems",
    "neural networks learn from training data",
    "the student reads a book",
    "the teacher explains the lesson",
    "the student learns machine learning",
    "natural language processing works with text",
    "transformers are useful for language tasks",
    "the cat and the dog are animals"
]

# Repeat the small corpus to provide more training examples
text = " ".join(sentences * 100)

# 3. Tokenize the text
tokenizer = Tokenizer(oov_token="<OOV>")
tokenizer.fit_on_texts([text])

token_ids = tokenizer.texts_to_sequences([text])[0]
vocab_size = len(tokenizer.word_index) + 1

print("Vocabulary size:", vocab_size)
print("Total tokens:", len(token_ids))

# 4. Create input-output pairs for next-token prediction
context_length = 8

X = []
y = []

for i in range(len(token_ids) - context_length):
    X.append(token_ids[i:i + context_length])
    y.append(token_ids[i + context_length])

X = np.array(X, dtype=np.int32)
y = np.array(y, dtype=np.int32)

# Shuffle and split into training and testing datasets
indices = np.random.permutation(len(X))
split_index = int(0.8 * len(X))

train_indices = indices[:split_index]
test_indices = indices[split_index:]

X_train, y_train = X[train_indices], y[train_indices]
X_test, y_test = X[test_indices], y[test_indices]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# 5. Build a Transformer-style model using self-attention
embedding_dim = 32
num_heads = 2
ff_dim = 64

inputs = Input(shape=(context_length,), dtype=tf.int32)

# Token embeddings
token_embeddings = Embedding(
    input_dim=vocab_size,
    output_dim=embedding_dim
)(inputs)

# Positional embeddings
positions = tf.range(start=0, limit=context_length, delta=1)
position_embeddings = Embedding(
    input_dim=context_length,
    output_dim=embedding_dim
)(positions)

x = token_embeddings + position_embeddings

# Basic causal self-attention
attention_output = MultiHeadAttention(
    num_heads=num_heads,
    key_dim=embedding_dim // num_heads
)(
    query=x,
    value=x,
    key=x,
    use_causal_mask=True
)

# Residual connection and normalization
x = LayerNormalization(epsilon=1e-6)(x + attention_output)

# Feed-forward network
ff_output = Dense(ff_dim, activation="relu")(x)
ff_output = Dense(embedding_dim)(ff_output)

# Second residual connection
x = LayerNormalization(epsilon=1e-6)(x + ff_output)

# Predict the next token
x = GlobalAveragePooling1D()(x)
x = Dropout(0.1)(x)
outputs = Dense(vocab_size, activation="softmax")(x)

model = Model(inputs=inputs, outputs=outputs)

# 6. Compile the model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nTransformer Model Summary:")
model.summary()

# 7. Train the model
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.1,
    verbose=1
)

# 8. Evaluate the model
test_loss, test_accuracy = model.evaluate(
    X_test, y_test, verbose=0
)

print("\nTest Loss:", round(float(test_loss), 4))
print("Test Accuracy:", round(float(test_accuracy) * 100, 2), "%")

# 9. Plot training and validation accuracy
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

# 10. Plot training and validation loss
plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.tight_layout()
plt.show()

# 11. Function to predict the next token
index_word = tokenizer.index_word

def predict_next_token(prompt):
    words = tokenizer.texts_to_sequences([prompt])[0]

    if len(words) == 0:
        print("Please enter a valid prompt.")
        return

    # Keep the most recent context_length tokens
    words = words[-context_length:]

    # Left-pad shorter inputs with zeros
    padded = [0] * (context_length - len(words)) + words
    input_array = np.array([padded], dtype=np.int32)

    probabilities = model.predict(input_array, verbose=0)[0]

    # Do not predict padding or the OOV token
    probabilities[0] = 0
    oov_id = tokenizer.word_index.get("<OOV>")
    if oov_id is not None:
        probabilities[oov_id] = 0

    next_id = int(np.argmax(probabilities))
    predicted_word = index_word.get(next_id, "<unknown>")

    print("Input text:", prompt)
    print("Predicted next token:", predicted_word)
    print("Prediction probability:", round(float(probabilities[next_id]), 4))

# 12. Test next-token prediction
print("\nNext-Token Prediction Examples:")
predict_next_token("the cat sits on the")
predict_next_token("machine learning is")
predict_next_token("a transformer model uses")

# 13. Generate text one token at a time
def generate_text(prompt, num_tokens=8):
    generated_words = prompt.lower().split()

    for _ in range(num_tokens):
        current_prompt = " ".join(generated_words[-context_length:])
        words = tokenizer.texts_to_sequences([current_prompt])[0]
        words = words[-context_length:]
        padded = [0] * (context_length - len(words)) + words

        probabilities = model.predict(
            np.array([padded], dtype=np.int32),
            verbose=0
        )[0]

        probabilities[0] = 0
        oov_id = tokenizer.word_index.get("<OOV>")
        if oov_id is not None:
            probabilities[oov_id] = 0

        next_id = int(np.argmax(probabilities))
        next_word = tokenizer.index_word.get(next_id)

        if next_word is None:
            break

        generated_words.append(next_word)

    return " ".join(generated_words)

print("\nGenerated Text:")
print(generate_text("the cat", num_tokens=8))

print("\nTransformer Next-Token Prediction Completed Successfully!")
```
